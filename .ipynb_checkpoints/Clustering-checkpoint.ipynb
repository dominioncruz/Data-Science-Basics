```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

from kneed import KneeLocator  # For detecting the elbow point

import warnings
warnings.filterwarnings('ignore')
```

```python
# Load the dataset
data = pd.read_csv("Wholesale customers data.csv")

# Select numerical features
features = data.drop(['Channel', 'Region'], axis=1)

# Standardize the features
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)
scaled_features
```

```python
# Calculate inertia for different K values
inertia = []
K = range(1, 11)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(scaled_features)
    inertia.append(kmeans.inertia_)
```

```python
# Plot the Elbow Graph
plt.figure(figsize=(8, 4))
plt.plot(K, inertia, 'bo-', markersize=8)
plt.xlabel('Number of clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method For Optimal K')
plt.xticks(K)
plt.show()
```

```python
# Use Kneedle algorithm to detect the elbow point
kneedle = KneeLocator(K, inertia, curve='convex', direction='decreasing')
optimal_k_elbow = kneedle.elbow
print(f'Optimal K according to the Elbow Method: {optimal_k_elbow}')
```

```python
# Calculate silhouette scores for different K values (from 2 to 10)
silhouette_scores = []
K_silhouette = range(2, 11)

for k in K_silhouette:
    kmeans = KMeans(n_clusters=k, random_state=42)
    cluster_labels = kmeans.fit_predict(scaled_features)
    silhouette_avg = silhouette_score(scaled_features, cluster_labels)
    silhouette_scores.append(silhouette_avg)
```

```python
# Plot Silhouette Scores
plt.figure(figsize=(8, 4))
plt.plot(K_silhouette, silhouette_scores, 'bo-', markersize=8)
plt.xlabel('Number of clusters (K)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Analysis For Optimal K')
plt.xticks(K_silhouette)
plt.show()
```

```python
# Find the maximum silhouette score
optimal_k_silhouette = K_silhouette[np.argmax(silhouette_scores)]
print(f'Optimal K according to the Silhouette Score: {optimal_k_silhouette}')
```

```python
print(f"Optimal K found using the Elbow Method: {optimal_k_elbow}")
print(f"Optimal K found using the Silhouette Score: {optimal_k_silhouette}")
```

```python
# Choose the optimal K (you can decide based on the methods above)
optimal_k = optimal_k_silhouette

# Apply K-Means with optimal K
kmeans_optimal = KMeans(n_clusters=optimal_k, random_state=42)
cluster_labels = kmeans_optimal.fit_predict(scaled_features)
```

```python
# Range for the number of clusters
K_range = range(2, 11)
silhouette_scores = []

for k in K_range:
    agg_clustering = AgglomerativeClustering(n_clusters=k, affinity='euclidean', linkage='ward')
    cluster_labels = agg_clustering.fit_predict(scaled_features)
    silhouette_avg = silhouette_score(scaled_features, cluster_labels)
    silhouette_scores.append(silhouette_avg)
```

```python
# Plot Silhouette Scores
plt.figure(figsize=(8, 4))
plt.plot(K_range, silhouette_scores, 'bo-', markersize=8)
plt.xlabel('Number of clusters (K)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Analysis For Optimal K')
plt.xticks(K_range)
plt.show()
```

```python
# Find the optimal K based on the highest silhouette score
optimal_k = K_range[np.argmax(silhouette_scores)]
print(f'Optimal K according to the Silhouette Score: {optimal_k}')
```

```python
#Define a Function to Apply DBSCAN and Calculate Metrics
def dbscan_clustering(eps_value, min_samples_value):
    dbscan = DBSCAN(eps=eps_value, min_samples=min_samples_value)
    labels = dbscan.fit_predict(scaled_features)
    
    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
    n_noise = list(labels).count(-1)
    
    if n_clusters > 1:
        silhouette_avg = silhouette_score(scaled_features, labels)
    else:
        silhouette_avg = -1  # Silhouette score is not meaningful for 1 cluster

    print(f'eps: {eps_value}, min_samples: {min_samples_value}, Estimated clusters: {n_clusters}, Noise points: {n_noise}, Silhouette Score: {silhouette_avg:.4f}')
    return labels, n_clusters, n_noise, silhouette_avg
```

```python
# Define parameter ranges
eps_values = [1.5, 1.8, 2.0, 2.2]
min_samples_values = [3, 5, 7]

# Store results
results = []

for eps in eps_values:
    for min_samples in min_samples_values:
        labels, n_clusters, n_noise, silhouette_avg = dbscan_clustering(eps, min_samples)
        results.append((eps, min_samples, n_clusters, n_noise, silhouette_avg))
```

```python
# Set optimal parameters
optimal_eps = 1.5
optimal_min_samples = 3

# Apply DBSCAN with optimal parameters
dbscan_optimal = DBSCAN(eps=optimal_eps, min_samples=optimal_min_samples)
cluster_labels = dbscan_optimal.fit_predict(scaled_features)

# Number of clusters and noise points
n_clusters_optimal = len(set(cluster_labels)) - (1 if -1 in cluster_labels else 0)
n_noise_optimal = list(cluster_labels).count(-1)

print(f'\nOptimal eps: {optimal_eps}, Optimal min_samples: {optimal_min_samples}')
print(f'Number of clusters: {n_clusters_optimal}')
print(f'Number of noise points: {n_noise_optimal}')
```

```python
# Apply PCA
pca = PCA(n_components=2)
principalComponents = pca.fit_transform(scaled_features)

# Create a DataFrame with the principal components and cluster labels
pca_df = pd.DataFrame(data=principalComponents, columns=['PC1', 'PC2'])
pca_df['Cluster'] = cluster_labels
pca_df
```

```python
# Map cluster labels to colors
unique_labels = set(cluster_labels)
colors = plt.cm.get_cmap('tab10', len(unique_labels))

# Plot the clusters
plt.figure(figsize=(8, 6))
for label in unique_labels:
    label_mask = (pca_df['Cluster'] == label)
    color = 'k' if label == -1 else colors(label)
    label_name = 'Noise' if label == -1 else f'Cluster {label}'
    plt.scatter(pca_df.loc[label_mask, 'PC1'], pca_df.loc[label_mask, 'PC2'],
                c=[color], label=label_name, edgecolors='none', alpha=0.7)

plt.title(f'DBSCAN Clustering Results (eps={optimal_eps}, min_samples={optimal_min_samples})')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.show()
```
