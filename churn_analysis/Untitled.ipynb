```python
data = pd.read_csv('Customer-Churn.csv')
```

```python
features = data[['gender', 'Dependents', 'MonthlyCharges', 'TotalCharges']]  #extract necessary features
```

```python
features
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
features['gender'] = le.fit_transform(features['gender'])
```

```python
features
```

```python
features['Dependents'] = le.fit_transform(features['Dependents'])
```

```python
features
```

```python
features['TotalCharges'] = le.fit_transform(features['TotalCharges'])
```

```python
features
```

```python
target = le.fit_transform(data['Churn'])
```

```python
target
```

```python
from sklearn.model_selection import train_test_split
```

```python
feature_train, feature_test, target_train, target_test = train_test_split(features, target, test_size = 0.3)
```

```python
print(feature_train.shape, feature_test.shape, target_train.shape, target_test.shape)
```

```python
feature_train
```

```python
from sklearn.svm import SVC
```

```python
classifier = SVC()
```

```python
feature_train.info()
```

```python
model = classifier.fit(feature_train, target_train)
```

```python
prediction = model.predict(feature_test)
```

```python
prediction
```

```python
diction = {"Actual  value": target_test, "Predictions": prediction}
```

```python
diction
```

```python
pd.DataFrame(diction)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(target_test, prediction)
```

```python
from sklearn.metrics import confusion_matrix
```

```python
print(confusion_matrix(target_test, prediction))
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
classifier2 = LogisticRegression()
```

```python
model2 = classifier2.fit(feature_train,target_train)
```

```python
prediction2 = model2.predict(feature_test)
```
