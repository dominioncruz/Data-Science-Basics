```python
features = data[['gender', 'Dependents', 'MonthlyCharges', 'TotalCharges']]  #extract necessary features
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
features['gender'] = le.fit_transform(features['gender'])
```

```python
features
```

```python
features['Dependents'] = le.fit_transform(features['Dependents'])
```

```python
features
```

```python
target = le.fit_transform(data['Churn'])
```

```python
target
```

```python
from sklearn.model_selection import train_test_split
```

```python
feature_train, feature_test, target_train, target_test = train_test_split(features, target, test_size = 0.3)
```
